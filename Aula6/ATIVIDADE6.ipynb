```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
t2012 = pd.read_csv("Tabela5-sem_emprego_2012.csv", sep=";", decimal=",")
t2026 = pd.read_csv("Tabela5-sem_emprego_2026.csv", sep=";", decimal=",")

t2012.head()
```

```python
# Junta as duas Tabelas 5 usando os identificadores das UFs
geral = t2012.merge(
    t2026,
    on=["Sigla", "Código", "Estado"],
    how="inner",
)

geral.head()
```

```python
# Importa a Tabela 1.1.1 utilizada na Aula 2
path_ibge = r"..\Aula2\Tabela 1.1.1.xls"
bruto = pd.read_excel(path_ibge, engine="xlrd", header=None, sheet_name="2022")

# Linhas 0 a 7: título/cabeçalho; 8 a 40: Brasil, regiões e UFs
ibge = bruto.iloc[8:41].copy()
ibge.columns = [
    "Estado",
    "horas_total_2022",
    "total_branca_2022",
    "total_preta_parda_2022",
    "homem_branca_2022",
    "homem_preta_parda_2022",
    "mulher_branca_2022",
    "mulher_preta_parda_2022",
]

regioes = ["Brasil", "Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]
ibge_estados = ibge[~ibge["Estado"].isin(regioes)].copy()
ibge_estados.head()
```

```python
# Cruza a comparação de 2012/2026 com as horas de cuidados domésticos de 2022
geral = geral.merge(ibge_estados, on="Estado", how="inner")

geral["Variação homens (p.p.)"] = (
    geral["Desocupados - homens (2026 T1)"]
    - geral["Desocupados - homens (2012 T1)"]
)
geral["Variação mulheres (p.p.)"] = (
    geral["Desocupados - mulheres (2026 T1)"]
    - geral["Desocupados - mulheres (2012 T1)"]
)

geral
```

```python
# Salva o resultado final
geral.to_csv("geral.csv", sep=";", decimal=",", index=False, encoding="utf-8-sig")
```

```python
# Comparação da composição da desocupação por sexo em 2012 e 2026
comparacao = geral[[
    "Estado",
    "Desocupados - homens (2012 T1)",
    "Desocupados - homens (2026 T1)",
    "Desocupados - mulheres (2012 T1)",
    "Desocupados - mulheres (2026 T1)",
]].set_index("Estado")

comparacao.plot(kind="bar", figsize=(16, 6))
plt.title("Composição da desocupação por sexo: 2012 T1 x 2026 T1")
plt.xlabel("Estado")
plt.ylabel("Percentual (%)")
plt.xticks(rotation=90)
plt.legend(title="Indicador", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```
