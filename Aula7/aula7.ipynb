```python
import pandas as pd

df_raw = pd.read_excel(path, sheet_name='BR GR UF MU', skiprows=1)

races = pd.Series(df_raw.columns).ffill().values
transports = df_raw.iloc[0].values

col_names = ['Localidade']
for r, t in zip(races[1:], transports[1:]):
  col_names.append(f'{r}_{t}')

df = df_raw.copy()
df.columns = col_names


for col in col_names[1:]:
  df[col] = pd.to_numeric(
      df[col].astype(str).str.replace(',', '.').str.strip(), errors='coerce'
  )

estados = df.iloc[7:34].copy()
cidades = df.iloc[34:].copy()

df_estados_melt = estados.melt(
    id_vars=['Localidade'],
    var_name='Raca_Transporte',
    value_name='Percentual',
)
df_estados_melt[['Raca', 'Transporte']] = df_estados_melt[
    'Raca_Transporte'
].str.split('_', expand=True)

df_cidades_melt = cidades.melt(
    id_vars=['Localidade'],
    var_name='Raca_Transporte',
    value_name='Percentual',
)
df_cidades_melt[['Raca', 'Transporte']] = df_cidades_melt[
    'Raca_Transporte'
].str.split('_', expand=True)

est_mais_coletivo = (
    df_estados_melt[df_estados_melt['Transporte'] == 'Transporte coletivo']
    .groupby('Localidade')['Percentual']
    .sum()
    .idxmax()
)

est_menos_coletivo = (
    df_estados_melt[df_estados_melt['Transporte'] == 'Transporte coletivo']
    .groupby('Localidade')['Percentual']
    .sum()
    .idxmin()
)

cid_mais_bike = (
    df_cidades_melt[df_cidades_melt['Transporte'] == 'Bicicleta']
    .groupby('Localidade')['Percentual']
    .sum()
    .idxmax()
)

cid_mais_carro = (
    df_cidades_melt[
        df_cidades_melt['Transporte'] == 'Automóvel, taxi ou assemelhados'
    ]
    .groupby('Localidade')['Percentual']
    .sum()
    .idxmax()
)

print('1. Estado com mais mulheres no transporte coletivo:', est_mais_coletivo)
print(
    '2. Estado com menos mulheres no transporte coletivo:', est_menos_coletivo
)
print('3. Cidade com mais mulheres de bicicleta:', cid_mais_bike)
print('4. Cidade com mais mulheres utilizando carro:', cid_mais_carro)
```
