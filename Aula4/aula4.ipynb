```python
#python -m pip install xlrd
import pandas as pd
```

```python
path = r"H:\python\3ETAPA\Aula4\CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"
```

```python
bruto = pd.read_excel(path, header = 2)
```

```python
tabela = bruto.iloc[:5570,1:].copy()
tabela
```

```python
TABELA_UF = (
    tabela.groupby("UF")[["População Município 2010\n(Sinopse)", "População Censo 2022"]].sum().reset_index())
TABELA_UF
```

```python
TABELA_UF["DIFERENÇA"] = TABELA_UF["População Censo 2022"] - TABELA_UF["População Município 2010\n(Sinopse)"]
TABELA_UF
```

```python
TABELA_UF.to_excel(r"H:\python\3ETAPA\Aula4\TABELADIFERENÇA.xlsx")
```

```python
TABELA_municipio = (
    tabela.groupby("NOME DO MUNICÍPIO")[["População Município 2010\n(Sinopse)", "População Censo 2022"]].sum().reset_index())
TABELA_municipio
```

```python
TABELA_municipio["DIFERENÇA"] = TABELA_municipio["População Censo 2022"] - TABELA_municipio["População Município 2010\n(Sinopse)"]
TABELA_municipio
```

```python
TABELA_municipio.to_excel(r"H:\python\3ETAPA\Aula4\TABELADIFERENÇA_MUNICIPIO.xlsx")
```
