```python
bruto = pd.read_excel(path, engine="openpyxl", header=None)
#header le sem usar a 1ª linha como nome de coluna (senão o título vira "coluna")

tabela = bruto.iloc[3:36].copy()
tabela.columns = [
    "Regiao/Estado",
    "Total_Pessoas",
    "Total_Homens",
    "Total_Mulheres",
    "Total_Ciência, Tecnologia, Engenharias e Matemática",
    "Total_Ciência, Tecnologia, Engenharias e Matemática_Homens",
    "Total_Ciência, Tecnologia, Engenharias e Matemática_Mulheres",
    "Total_Educação, Serviços pessoais, Saúde e Bem-estar",
    "Total_Educação, Serviços pessoais, Saúde e Bem-estar_Homens",
    "Total_Educação, Serviços pessoais, Saúde e Bem-estar_Mulheres"
]
tabela
```

```python
#Transforma os dados em números (para pode fazer calculos)

for col in tabela.columns[1:]:
    tabela[col] = pd.to_numeric(tabela[col])
```

```python
regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]

mascara_regioes = tabela["Regiao/Estado"].isin(regioes)

Total_HM_Superior_Regiao = tabela.loc[
    mascara_regioes,
    ["Regiao/Estado", "Total_Homens", "Total_Mulheres"]
]
Total_HM_Superior_Regiao
```

```python
Total_HM_CienciasEtc = tabela.loc[
    mascara_regioes,
    ["Regiao/Estado","Total_Ciência, Tecnologia, Engenharias e Matemática_Homens","Total_Ciência, Tecnologia, Engenharias e Matemática_Mulheres"]
]

Total_HM_CienciasEtc
```

```python
Total_HM_EducacaoEtc = tabela.loc[
    mascara_regioes,
    ["Regiao/Estado","Total_Educação, Serviços pessoais, Saúde e Bem-estar_Homens","Total_Educação, Serviços pessoais, Saúde e Bem-estar_Mulheres"]
]
Total_HM_EducacaoEtc
```
